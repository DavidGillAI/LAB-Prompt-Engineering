```python
from openai import OpenAI
from dotenv import load_dotenv
import os
import time

load_dotenv()

client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

print("Setup complete. OpenAI client is ready.")
```

```python
def get_completion(prompt, model="gpt-4o-mini"):
    response = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "user", "content": prompt}
        ],
        temperature=0.7
    )
    
    return response.choices[0].message.content


print("Helper function ready.")
```

```python
def run_multiple_tests(prompt, runs=5):
    results = []
    
    for i in range(runs):
        response = get_completion(prompt)
        results.append(response)
        
    return results


print("Testing function ready.")
```

```python
# Initial simple prompt for sentiment analysis

sentiment_prompt_v1 = """
Classify this customer message: "I love this product! It's exactly what I needed."
"""

# Test it once
result = get_completion(sentiment_prompt_v1)

print("Sentiment Analysis Result:")
print(result)
```

```python
# Initial simple prompt for product description

product_prompt_v1 = """
Create a product description for a wireless mouse that costs $29.99.
"""

# Test it once
result = get_completion(product_prompt_v1)

print("Product Description Result:")
print(result)
```

```python
# Initial simple prompt for data extraction

extraction_prompt_v1 = """
Extract information from this customer feedback: "I ordered item #12345 on March 15th. The delivery was fast but the packaging was damaged."
"""

# Test it once
result = get_completion(extraction_prompt_v1)

print("Data Extraction Result:")
print(result)
```

```python
sentiment_prompt_v3 = """
Classify the sentiment of the customer message as Positive, Negative, or Neutral.

Respond with only one word.

Examples:
Customer message: "This product works perfectly and arrived quickly."
Sentiment: Positive

Customer message: "The item broke after one day and support was unhelpful."
Sentiment: Negative

Customer message: "The package arrived yesterday."
Sentiment: Neutral

Customer message: "I love this product! It's exactly what I needed."
Sentiment:
"""

result = get_completion(sentiment_prompt_v3)

print("Few-Shot Sentiment Result:")
print(result)
```

### Sentiment Prompt V3 Observation

After adding few-shot examples and running the prompt 15 times, all outputs returned the single-word label `Positive`. This shows a major improvement compared with version 1, where the model used inconsistent wording, formatting, and label names. The few-shot examples helped enforce the expected output format and label vocabulary.

```python
extraction_prompt_v3 = """
Extract structured information from the customer feedback.

First, think through the feedback step by step:
1. Identify the order number.
2. Identify the order date.
3. Identify the delivery speed.
4. Identify the packaging condition.

Then return the final answer in this exact JSON format:

{
  "order_number": "",
  "order_date": "",
  "delivery_speed": "",
  "packaging_condition": ""
}

Customer feedback:
"I ordered item #12345 on March 15th. The delivery was fast but the packaging was damaged."
"""

result = get_completion(extraction_prompt_v3)

print("Chain-of-Thought Data Extraction Result:")
print(result)
```

```python
extraction_v3_results_15 = run_multiple_tests(extraction_prompt_v3, runs=15)

print("Chain-of-Thought Data Extraction - 15 Runs")

for i, result in enumerate(extraction_v3_results_15, start=1):
    print(f"\nRun {i}:")
    print(result)
```

### Data Extraction Prompt V3 Observation

After adding Chain-of-Thought reasoning and testing 15 times, the extracted values were consistent across all runs. However, the output format still varied. Some responses returned clean JSON only, while others included reasoning steps, introductory text, or JSON inside markdown code blocks. This shows that Chain-of-Thought improved extraction reliability, but did not fully solve structured output consistency.

### Product Description Prompt V3 Observation

After combining few-shot examples with structured formatting instructions, the product descriptions became significantly more consistent in structure, tone, and length across 15 runs. Most responses followed the same layout with a short introduction, exactly 3 bullet points, and a short call to action.

However, some hallucinated claims still appeared, including references to ergonomic comfort, compatibility, and connectivity features that were not explicitly provided in the prompt. This shows that few-shot prompting improved formatting consistency more than factual grounding.

```python
sentiment_variations = [
    "This is the worst purchase I've ever made.",
    "The package arrived this morning.",
    "Customer support was helpful and solved my issue quickly."
]

print("Sentiment Variation Tests:\n")

for message in sentiment_variations:
    
    prompt = f"""
Classify the sentiment of the customer message as Positive, Negative, or Neutral.

Respond with only one word.

Examples:
Customer message: "This product works perfectly and arrived quickly."
Sentiment: Positive

Customer message: "The item broke after one day and support was unhelpful."
Sentiment: Negative

Customer message: "The package arrived yesterday."
Sentiment: Neutral

Customer message: "{message}"
Sentiment:
"""

    result = get_completion(prompt)

    print(f"Message: {message}")
    print(f"Sentiment: {result}\n")
```

### Sentiment Variation Observation

The few-shot sentiment prompt performed consistently across multiple customer message variations. All responses used the correct one-word format and correctly classified positive, negative, and neutral examples. This suggests the prompt generalises well to similar sentiment analysis tasks.

## Final Evaluation and Comparison Report

| Task | Version 1 Issues | Version 3 Improvements | Remaining Issues |
|---|---|---|---|
| Sentiment Analysis | Inconsistent labels, formatting differences, multiple response styles | Few-shot examples and strict instructions produced highly consistent one-word outputs across 15 runs | Very few remaining issues observed |
| Product Description Generation | Large formatting variation, excessive hallucinated product specifications, inconsistent structure | Few-shot examples and structure constraints created highly consistent formatting, tone, and length | Some hallucinated marketing claims still appeared, including ergonomic and compatibility features not provided in the prompt |
| Data Extraction | Formatting inconsistencies, inconsistent order number formatting, introductory text variations | Chain-of-Thought prompting improved extraction consistency and produced reliable structured data | Some runs still included reasoning text and markdown code blocks instead of clean JSON-only output |

### Overall Findings

The lab demonstrated that prompt engineering techniques such as explicit constraints, few-shot prompting, structured formatting, and Chain-of-Thought reasoning can significantly improve consistency and reliability in LLM outputs.

Few-shot prompting was especially effective for enforcing output style and formatting consistency. Chain-of-Thought reasoning improved extraction reliability, but sometimes introduced additional explanatory text that reduced JSON formatting consistency.

The final prompts performed substantially better than the original zero-shot prompts, especially for sentiment classification and structured data extraction.
